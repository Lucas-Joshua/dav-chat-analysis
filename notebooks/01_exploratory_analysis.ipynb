In this notebook we explore the WhatsApp chat dataset.
The goal is to inspect structure, message distribution and basic statistics.

```python
import sys

sys.path.append("..")

from src import config
from src.modules import data_loader, data_cleaning
```

```python
df_raw = data_loader.load_raw_chat(config.RAW_DATA_FILE)
df_raw.head()
```

```python
df_clean.describe()
```

```python
df_clean["sender"].value_counts().head(10)
```

```python
import matplotlib.pyplot as plt

top_users = df_clean["sender"].value_counts().head(10)

plt.figure(figsize=(8, 6))
top_users.sort_values().plot(kind="barh")

plt.title("Top 10 most active users")
plt.xlabel("Number of messages")
plt.tight_layout()
plt.show()
```
